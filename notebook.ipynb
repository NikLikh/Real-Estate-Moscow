```python
from playwright.async_api import async_playwright
import pandas as pd
import atexit
import asyncio
```

```python
data = []

async def scrape(playwright):
    browser = await playwright.chromium.launch(headless=False)
    page = await browser.new_page()

    await page.goto('https://cian.ru/') 

    urls = await page.query_selector_all('test')
    for url in urls:
        href = await url.get_attribute('href')
        data.append({'url': href})

    titles = await page.query_selector_all('test')
    for i, title in enumerate(titles):
        data[i]['title'] = await title.inner_text()

    await browser.close()
    await playwright.stop()

playwright = async_playwright().start()
scrape(playwright)

df = pd.DataFrame(data)
print(df)
```
